# Data Exploration
Visualize samples from both datasets before training.

```python
import os, sys
sys.path.insert(0, os.path.join(os.getcwd(), '..'))
import config
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import random
import glob
import numpy as np
```

```python
# ── Classification dataset samples ──────────────────────────────────────────
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.flatten()

for i, class_name in enumerate(config.CLASS_NAMES):
    class_dir = os.path.join(config.CLASSIFICATION_DATA_DIR, 'Training', class_name)
    imgs = glob.glob(os.path.join(class_dir, '*.jpg'))
    if not imgs:
        continue
    sample = random.choice(imgs)
    img = mpimg.imread(sample)
    axes[i].imshow(img, cmap='gray' if img.ndim == 2 else None)
    axes[i].set_title(class_name.capitalize(), fontsize=13, fontweight='bold')
    axes[i].axis('off')

    # Distribution bar (second row)
    count = len(imgs)
    axes[i+4].bar([class_name], [count], color='steelblue')
    axes[i+4].set_title(f'{count} images')

fig.suptitle('Classification Dataset — Sample Images & Counts', fontsize=15)
plt.tight_layout()
plt.show()
```

```python
# ── Segmentation dataset samples ────────────────────────────────────────────
mask_paths = glob.glob(os.path.join(config.SEGMENTATION_DATA_DIR, '**', '*_mask.tif'), recursive=True)[:4]

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for i, msk_path in enumerate(mask_paths):
    img_path = msk_path.replace('_mask.tif', '.tif')
    if not os.path.exists(img_path):
        continue
    img = mpimg.imread(img_path)
    msk = mpimg.imread(msk_path)
    axes[0, i].imshow(img, cmap='gray' if img.ndim == 2 else None)
    axes[0, i].set_title('MRI Slice')
    axes[0, i].axis('off')
    axes[1, i].imshow(msk, cmap='Reds')
    axes[1, i].set_title('Ground Truth Mask')
    axes[1, i].axis('off')

fig.suptitle('Segmentation Dataset — MRI Slices & Ground Truth Masks', fontsize=15)
plt.tight_layout()
plt.show()
```
